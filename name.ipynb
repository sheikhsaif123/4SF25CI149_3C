```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Iris.csv")
print(df.shape)
```

```python
print(df.head(5))
```

```python
print(df.info())
```

```python
df['Species'] = df['Species'].astype('category')
```

```python
print(df.dtypes)
```

```python
df['PetalArea'] = df['PetalLengthCm'] * df['PetalWidthCm']
```

```python
print(df.head(5))
```

```python

print(df.isnull().sum())
```

```python
random_rows = np.random.choice(df.index, size=10, replace=False)
df.loc[random_rows, 'SepalLengthCm'] = np.nan
```

```python
print(df.isnull().sum())
```

```python
   
df['SepalLengthCm'] = df['SepalLengthCm'].fillna(
    df['SepalLengthCm'].mean()
) 
```

```python
print(df.isnull().sum())
```

```python
print("\nSpecies Count:")
print(df['Species'].value_counts())
```

```python
print("\nAverage Sepal Length by Species:")
print(df.groupby('Species', observed=True)['SepalLengthCm'].mean())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

sns.countplot(data=df, x='Species')
plt.title("Count of Each Iris Species")
plt.show()
```

```python

plt.hist(df['SepalLengthCm'], bins=15)
plt.title("Sepal Length Distribution")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```
